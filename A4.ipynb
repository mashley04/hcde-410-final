# Data — the dataset(s), what they contain, a link, the license/terms, why they fit, and any ethical considerations.
I will be using the dataset "Impact of AI on students" from https://www.kaggle.com/datasets/laveshjadon/ai-impact-on-students?resource=download. 

This dataset includes:
- Unique ID for identifying students
- Academic Major
- Year of study
- Pre-semester GPA
- Weekly hour AI use
- Main purpose for using AI
- Self rated proficiency in using AI
- Number of distinct AI tools used
- Whether they have a paid subscription or not
- Students study hours without AI
- Percieved AI dependancy
- Anxiety Level during exams
- Post-Semester GPA
- Skill Retention score
- Burn out risk level

This dataset has all the necessary information I need to analyze how students engage with AI based on their demographic and how AI usage affects their studies.
